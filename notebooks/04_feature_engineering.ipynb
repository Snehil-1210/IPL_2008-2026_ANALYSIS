### 1. Mathematical Formulas & Methodology

#### A. Batting Performance Metrics
- **Batting Average**: $\text{Avg} = \frac{\text{Total Runs}}{\text{Dismissals}}$ (Handles not-outs without division by zero).
- **Batting Strike Rate**: $\text{SR} = \frac{\text{Runs}}{\text{Balls Faced}} \times 100$ (Excludes wides from balls faced).
- **Boundary Percentage**: $\text{Boundary \%} = \frac{4 \times \text{Fours} + 6 \times \text{Sixes}}{\text{Total Runs}} \times 100$.
- **Phase Breakdown**: Powerplay (Overs 1–6), Middle (Overs 7–15), Death (Overs 16–20).
- **Batting Impact Score**: Comprehensive index evaluating volume, scoring acceleration, and boundary pressure:
  $$\text{Batting Impact} = (\text{Runs / Inning} \times 0.4) + \left(\frac{\text{Strike Rate}}{100} \times 20\right) + (\text{Boundary \%} \times 0.5)$$

#### B. Bowling Performance Metrics
- **Economy Rate**: $\text{Econ} = \frac{\text{Runs Conceded}}{\text{Legal Balls Bowled} / 6.0}$ (Excludes byes/leg-byes; excludes wides/noballs from legal balls).
- **Bowling Average**: $\text{Bowl Avg} = \frac{\text{Runs Conceded}}{\text{Bowler Wickets}}$ (Excludes non-bowler wickets like run outs).
- **Bowling Strike Rate**: $\text{Bowl SR} = \frac{\text{Legal Balls}}{\text{Bowler Wickets}}$.
- **Dot Ball Percentage**: $\text{Dot \%} = \frac{\text{Dot Balls}}{\text{Total Deliveries}} \times 100$.
- **Bowling Impact Score**:
  $$\text{Bowling Impact} = (\text{Wickets / Inning} \times 25) + \max(0, 8.5 - \text{Economy}) \times 8.0 + (\text{Dot Ball \%} \times 0.4)$$

#### C. Team & Match Analytics
- **Win Percentage**: $\text{Win \%} = \frac{\text{Wins}}{\text{Matches}} \times 100$.
- **Chase Success Rate**: Wins Batting 2nd / Total Matches Chasing $\times 100$.
- **Defend Success Rate**: Wins Batting 1st / Total Matches Defending $\times 100$.
- **Toss-to-Match Conversion Rate**: Matches Won When Winning Toss / Total Toss Wins $\times 100$.
