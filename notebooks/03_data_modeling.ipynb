### 1. Objective & Architecture
In high-performance business intelligence (Power BI / DAX), flat tables cause severe performance bottlenecks, cross-filter ambiguity, and memory bloat. We convert our cleaned datasets into a formal **Star Schema** with surrogate keys (`1:*` single-direction relationships):

```
       +--------------+
       |   Dim_Date   |
       +-------+------+
               |
               v [date_key]
+--------------+------+       [venue_key]       +--------------+
|     Fact_Matches    +------------------------>|   Dim_Venue  |
+--------------+------+                         +--------------+
       |       |
       |       +-------------> [team1_key, team2_key, winner_key] 
       |                                        |
       v [match_key]                            v
+--------------+------+                         +--------------+
|   Fact_BallByBall   +------------------------>|   Dim_Team   |
+--------------+------+                         +--------------+
               |
               +-------------> [striker_key, bowler_key] 
                                                |
                                                v
                                        +--------------+
                                        |  Dim_Player  |
                                        +--------------+
```
