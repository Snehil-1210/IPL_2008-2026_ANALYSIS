### 2. Export Inventory
The tables in `data/processed/` form our enterprise analytical star schema:
- `dim_team.csv`: 15 franchises with unique `team_key` and short codes.
- `dim_player.csv`: 808 unique players with `player_key`.
- `dim_venue.csv`: 36 stadiums with city, country, and `venue_key`.
- `dim_date.csv`: Full date dimension with calendar attributes.
- `fact_matches.csv`: 1,243 match outcomes with surrogate keys.
- `fact_ballbyball.csv`: 295,732 delivery-level records.
- `agg_batting_career.csv`: Career batting metrics and impact indices.
- `agg_bowling_career.csv`: Career bowling metrics and impact indices.
- `agg_team_overall.csv`: Franchise historical win and toss rates.
- `agg_venue_stats.csv`: Stadium pitch profiles and chasing bias.

```python
import os
import pandas as pd

processed_dir = '../data/processed'
files = [f for f in os.listdir(processed_dir) if f.endswith('.csv')]

inventory = []
for f in sorted(files):
    path = os.path.join(processed_dir, f)
    df = pd.read_csv(path, nrows=1)
    row_count = sum(1 for _ in open(path, 'r', encoding='utf-8', errors='ignore')) - 1
    size_mb = os.path.getsize(path) / (1024 * 1024)
    inventory.append({
        'File Name': f,
        'Row Count': f"{row_count:,}",
        'Columns': len(df.columns),
        'Size (MB)': f"{size_mb:.2f} MB"
    })

df_inv = pd.DataFrame(inventory)
display(df_inv)
```
